## iii. Chuyển ETH: account #3 → account #6

### Các bước chương trình
1. Kết nối RPC Ganache.
2. Lấy `private_key` + `address` người gửi (#3), `address` người nhận (#6).
3. `amount = to_wei(...)`; lấy `nonce`.
4. Tạo dict `transaction` (`from`, `to`, `value`, `gas`, `gasPrice`, `nonce`).
5. `sign_transaction` → `send_raw_transaction` → (tuỳ chọn) chờ receipt.
6. In balance trước/sau để đối chiếu trên Ganache.


#### Bước 1 — Kết nối Ganache + khai báo account

````python
# transfer — account INDEX 3 -> INDEX 6
from web3 import Web3

w3 = Web3(Web3.HTTPProvider("http://127.0.0.1:7545"))
assert w3.is_connected(), "Mo Ganache truoc (http://127.0.0.1:7545)"

sender_private_key = "0x6552742db604c1cad2ebda4f5ba15a777999581d3a179d7e6c4a0b0950f61601"
sender_address = Web3.to_checksum_address("0x0527609ad5e731bb3752821063D6f57c7602cbbE")  # INDEX 3
recipient_address = Web3.to_checksum_address("0xb3233773DfB70B47f5Af0702392d84A1bE9Ed67F")  # INDEX 6

# De bai: 5 wei. Hinh 7: doi thanh Web3.to_wei(5, "ether")
amount = Web3.to_wei(5, "wei")
print("connected:", w3.is_connected())
print("sender   :", sender_address)
print("recipient:", recipient_address)
print("amount   :", amount, "wei")
````

#### Bước 2 — Balance **trước** khi chuyển

````python
print("=== BEFORE ===")
print("sender   :", w3.from_wei(w3.eth.get_balance(sender_address), "ether"), "ETH")
print("recipient:", w3.from_wei(w3.eth.get_balance(recipient_address), "ether"), "ETH")
````

#### Bước 3 — Tạo transaction (`gas`, `gasPrice`, `nonce`)

````python
transaction = {
    "from": sender_address,
    "to": recipient_address,
    "value": amount,
    "gas": 21000,  # transfer thuong
    "gasPrice": Web3.to_wei("50", "gwei"),
    "nonce": w3.eth.get_transaction_count(sender_address),
}

print("=== TRANSACTION ===")
for k, v in transaction.items():
    print(f"  {k}: {v}")
````

#### Bước 4 — Ký tx + gửi lên chain

````python
signed = w3.eth.account.sign_transaction(transaction, sender_private_key)
raw = getattr(signed, "raw_transaction", None) or signed.rawTransaction
tx_hash = w3.eth.send_raw_transaction(raw)
receipt = w3.eth.wait_for_transaction_receipt(tx_hash)
print("tx hash:", tx_hash.hex())
print("status :", receipt["status"], "(1=success)")
print("gasUsed:", receipt["gasUsed"])
````

#### Bước 5 — Balance **sau** khi chuyển + phí gas

````python
print("=== AFTER ===")
print("tx hash :", tx_hash.hex())
print("status  :", receipt["status"], "(1=success)")
print("gasUsed :", receipt["gasUsed"])
print("phi uoc tinh (wei):", receipt["gasUsed"] * transaction["gasPrice"])
print("sender   :", w3.from_wei(w3.eth.get_balance(sender_address), "ether"), "ETH")
print("recipient:", w3.from_wei(w3.eth.get_balance(recipient_address), "ether"), "ETH")
````
